```python
from typing import TypedDict , Annotated
import operator 
from langgraph.graph import StateGraph, START, END
```

```python
class ProfileState(TypedDict):
    username:str
    age:str
    is_adult:bool
    tags:Annotated[list[str],operator.add]
```

```python
def formet_profile(state:ProfileState):
    clean_name=state["username"].strip().title()
    return{
        "username":clean_name,
        "tags":["registered_User"]
    }
```

```python
def verify_age(state:ProfileState):
    user_age=state['age']
    adult_status=user_age>=18

    age_tag="adult_number" if adult_status else "minor_member"

    return{
        "is_adult":adult_status,
        "tags":[age_tag]
    }
```

```python
builder=StateGraph(ProfileState)
```

```python
builder.add_node("formetter",formet_profile)
builder.add_node("verifier",verify_age)

builder.add_edge(START,"formetter")
builder.add_edge("formetter","verifier")
builder.add_edge("verifier",END)
```

```python
graph=builder.compile()
```

```python
input_data={
    "username":"  ZAID saifi ",
    "age":12,
    "tags":["new_signup"]
}
```

```python
final_profile=graph.invoke(input_data)
print(final_profile)
```
