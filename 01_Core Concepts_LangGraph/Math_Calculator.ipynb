```python
from typing import TypedDict , Annotated
import operator 
from langgraph.graph import StateGraph,START,END
```

```python
class MathState(TypedDict):
    number:int
    history:Annotated[list[str],operator.add]
```

```python
def add_ten(State:MathState):
    new_val=State['number']+10
    return {
        "number":new_val,
        "history":[f"add 10 ,result is {new_val}"]

    }
```

```python
def multiply_by_two(State:MathState):
    new_val=State["number"]*2
    return{
        "number":new_val,
        "history":[f"multipled by 2, result is {new_val}"]
    }
```

```python
builder=StateGraph(MathState)
builder.add_node("adder",add_ten)
builder.add_node("multipier",multiply_by_two)
```

```python
builder.add_edge(START,"adder")
builder.add_edge("adder","multipier")
builder.add_edge("multipier",END)
```

```python
graph=builder.compile()
```

```python
# 5. Graph run karo
result = graph.invoke({
    "number": 5,
    "history": ["Started with 5"]
})

print(result)
```
