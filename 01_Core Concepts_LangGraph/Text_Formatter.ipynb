```python
from typing import TypedDict , Annotated
import operator
from langgraph.graph import StateGraph , START,END
```

```python
class MyState(TypedDict):
    input_text:str
    processed_text:str
    logs:Annotated[list[str],operator.add]
```

```python
# Node
def processed_node(state:MyState):
    raw_text=state["input_text"]
    result=raw_text.upper()

    return{
        "processed_text":result,
        "logs":["text is converted to uppercase"]
    }
```

```python
# Graph
builder=StateGraph(MyState)

# node
builder.add_node("processer",processed_node)

# Edges 
builder.add_edge(START,"processer")
builder.add_edge("processer",END)
```

```python
# compile 
graph=builder.compile()
```

```python
initial_input={
    "input_text":"HEllo langGraph",
    "logs":["Graph started"]
}

final_state=graph.invoke(initial_input)
```
