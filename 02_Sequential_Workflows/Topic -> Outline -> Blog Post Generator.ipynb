```python
from typing import TypedDict
from langgraph.graph import StateGraph , START,END
from langchain_groq import ChatGroq
```

```python

llm = ChatGroq(model="openai/gpt-oss-120b",api_key="X")
```

```python
class BlogState(TypedDict):
    topic:str
    outline:str
    blog_draft:str
    seo_tag:str
```

```python
def create_outline(state:BlogState):
    text=state['topic']
    prompt = f'Create a 3-point outline for a blog on: {text}'
    response=llm.invoke(prompt)
    return {"outline":response.content}
```

```python
def write_blog(state:BlogState):
    blog=state['outline']
    prompt = f'Write a short 100-word blog post using this outline: {blog}'
    response=llm.invoke(prompt)
    return {'blog_draft':response.content}
```

```python
def genrate_sso(state:BlogState):
    tag=state['blog_draft']
    prompt = f'Generate 5 SEO-friendly hashtags for this blog post: {tag}'
    response=llm.invoke(prompt)
    return {'seo_tag':response.content}
```

```python
builder=StateGraph(BlogState)
```

```python
builder.add_node("Outliner",create_outline)
builder.add_node("writter",write_blog)
builder.add_node("seo_taags",genrate_sso)

builder.add_edge(START,"Outliner")
builder.add_edge("Outliner","writter")
builder.add_edge("writter","seo_taags")
builder.add_edge("seo_taags",END)
```

```python
graph=builder.compile()
```

```python
result=graph.invoke({"topic":"Why do we judge people so quickly?"})
print(result["outline"])
print(result["blog_draft"])
print(result["seo_tag"])
```
